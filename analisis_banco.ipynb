```python
import pandas as pd

# 1. Cargamos el archivo de 2014
df_2014 = pd.read_csv('customer-details-2014.csv', sep=';', index_col=0)

# 2. Cargamos los anteriores (asegúrate de que los nombres sean correctos)
df_2012 = pd.read_csv('customer-details-2012.csv', sep=';', index_col=0)
df_2013 = pd.read_csv('customer-details-2013.csv', sep=';', index_col=0)

# 3. Unimos los tres años en una sola tabla de clientes
df_clientes = pd.concat([df_2012, df_2013, df_2014], ignore_index=True)

# 4. Cargamos también la tabla del banco para tenerla lista
df_banco = pd.read_csv('bank-additional.csv', sep=';', index_col=0)

print("--- TODO CARGADO Y UNIFICADO ---")
print(f"Total de registros de clientes (2012+2013+2014): {len(df_clientes)}")
print(f"Total de registros en la tabla del banco: {len(df_banco)}")
```

```python
# 1. Realizamos la unión (merge) de las dos tablas
# Usamos 'left_on' para la columna del banco y 'right_on' para la de clientes
df_final = pd.merge(df_banco, df_clientes, left_on='id_', right_on='ID')

# 2. Como ahora tenemos dos columnas de ID que son iguales, borramos una para no repetir
df_final = df_final.drop(columns=['ID'])

# 3. Revisamos el resultado final
print("--- UNIÓN COMPLETADA ---")
print(f"Ahora tienes una tabla maestra con {df_final.shape[0]} filas y {df_final.shape[1]} columnas.")
print("\nPrimeras 5 filas de tu nueva tabla unificada:")
display(df_final.head())

# 4. Tip extra: ¿Cuántos datos nos faltan ahora?
print("\nResumen de valores vacíos en la tabla final:")
print(df_final.isnull().sum())
```

```python
# 1. Identificar cuántos valores nulos hay por columna
print("Valores nulos antes de limpiar:")
print(df_final.isnull().sum())

# 2. LIMPIEZA DE 'Income' (Ingresos)
# Vamos a rellenar los ingresos vacíos con la MEDIANA (el valor central) 
# para que los valores extremos no afecten el análisis.
mediana_ingresos = df_final['Income'].median()
df_final['Income'] = df_final['Income'].fillna(mediana_ingresos)

# 3. LIMPIEZA DE 'age' (Edad)
# Para la edad, usaremos el promedio (mean) y lo convertiremos a número entero
promedio_edad = df_final['age'].mean()
df_final['age'] = df_final['age'].fillna(promedio_edad).astype(int)

# 4. NORMALIZAR TEXTO (Columna 'job' y 'marital')
# Pasamos todo a minúsculas para que 'MARRIED' y 'married' sean lo mismo
df_final['job'] = df_final['job'].str.lower()
df_final['marital'] = df_final['marital'].str.lower()

# 5. CONVERTIR FECHAS
# Transformamos 'Dt_Customer' a un formato de fecha real que Python entienda
df_final['Dt_Customer'] = pd.to_datetime(df_final['Dt_Customer'])

print("\n--- ¡LIMPIEZA COMPLETADA! ---")
print(f"Nulos restantes en Income: {df_final['Income'].isnull().sum()}")
print(f"Nulos restantes en age: {df_final['age'].isnull().sum()}")
display(df_final.head())
```

```python
import pandas as pd
import numpy as np

# --- 1. CARGA DE DATOS (Raw Data) ---
# Cargamos el banco y los 3 archivos de clientes
df_bank = pd.read_csv('bank-additional.csv', sep=';', index_col=0)
df_2012 = pd.read_csv('customer-details-2012.csv', sep=';', index_col=0)
df_2013 = pd.read_csv('customer-details-2013.csv', sep=';', index_col=0)
df_2014 = pd.read_csv('customer-details-2014.csv', sep=';', index_col=0)

# --- 2. CONCATENACIÓN Y UNIÓN (Merge) ---
# Unimos los clientes en un solo bloque
df_customers = pd.concat([df_2012, df_2013, df_2014], ignore_index=True)

# Unión maestra: usamos 'id_' del banco e 'ID' de clientes
df_raw = pd.merge(df_bank, df_customers, left_on='id_', right_on='ID', how='inner')

# --- 3. LIMPIEZA DE DATOS ---
# A. Eliminar duplicados si existen
df_clean = df_raw.drop_duplicates()

# B. Eliminar columnas redundantes (ID repetido)
df_clean = df_clean.drop(columns=['ID'])

# C. Gestión de Nulos (Imputación)
# - Ingresos: Usamos la mediana (más robusta frente a valores extremos)
df_clean['Income'] = df_clean['Income'].fillna(df_clean['Income'].median())
# - Edad: Usamos la media y convertimos a entero
df_clean['age'] = df_clean['age'].fillna(df_clean['age'].mean()).astype(int)

# D. Estandarización de texto
# Ponemos profesiones y estado civil en minúsculas y quitamos puntos finales
df_clean['job'] = df_clean['job'].str.lower().str.replace('.', '', regex=False)
df_clean['education'] = df_clean['education'].str.replace('.', '_', regex=False)

# E. Conversión de tipos
df_clean['Dt_Customer'] = pd.to_datetime(df_clean['Dt_Customer'])

# F. Feature Engineering (Creación de nuevas variables útiles para el análisis)
# Creamos una columna que sume niños y adolescentes para saber el total de menores en casa
df_clean['total_minors'] = df_clean['Kidhome'] + df_clean['Teenhome']

# --- 4. GUARDAR RESULTADOS ---
# El proyecto pide una carpeta de datos procesados. Guardamos el CSV limpio.
df_clean.to_csv('datos_limpios.csv', index=False)

print("¡Proceso de limpieza finalizado!")
print(f"Registros originales: {len(df_raw)} | Registros tras limpieza: {len(df_clean)}")
print(f"Nulos totales detectados: {df_clean.isnull().sum().sum()}")
```

```python
# --- 1. ESTADÍSTICAS GENERALES ---
# El método .describe() nos da media, mediana (50%), desviación estándar, etc.
print("--- Resumen Estadístico de Variables Numéricas ---")
stats_numericas = df_final[['age', 'Income', 'campaign', 'duration', 'NumWebVisitsMonth']].describe()
display(stats_numericas)

# --- 2. ANÁLISIS DE LA VARIABLE OBJETIVO (Tasa de Éxito) ---
# ¿Qué porcentaje de clientes aceptó el producto ('yes')?
print("\n--- Tasa de suscripción (Variable 'y') ---")
tasa_exito = df_final['y'].value_counts(normalize=True) * 100
print(tasa_exito)

# --- 3. ANÁLISIS POR SEGMENTOS (Agrupamientos) ---
# A. ¿Qué profesiones tienen el ingreso promedio más alto?
print("\n--- Top 5 Profesiones por Ingreso Promedio ---")
ingreso_por_job = df_final.groupby('job')['Income'].mean().sort_values(ascending=False)
print(ingreso_por_job.head())

# B. ¿Influye el estado civil en la aceptación del producto?
print("\n--- Tasa de éxito por Estado Civil ---")
exito_marital = df_final.groupby('marital')['y'].value_counts(normalize=True).unstack()['yes'] * 100
print(exito_marital.sort_values(ascending=False))

# --- 4. ANÁLISIS DE CORRELACIÓN ---
# Esto ayuda a ver si hay relación entre variables (ej. ¿más ingresos = más visitas web?)
print("\n--- Matriz de Correlación (relación con Income) ---")
# Seleccionamos solo las columnas numéricas para evitar errores
columnas_num = df_final.select_dtypes(include=['number'])
correlacion = columnas_num.corr()
# Mostramos solo cómo se relaciona todo con el ingreso
print(correlacion['Income'].sort_values(ascending=False))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Preparación de los datos para la visualización
# Calculamos la tasa de éxito (porcentaje de 'yes') por cada tipo de trabajo
# Usamos normalize=True para obtener proporciones y multiplicamos por 100
tasa_exito_job = df_clean.groupby('job')['y'].value_counts(normalize=True).unstack()['yes'] * 100
tasa_exito_job = tasa_exito_job.sort_values(ascending=False)

# 2. Configuración del estilo del gráfico
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 6))

# 3. Creación del gráfico de barras
sns.barplot(x=tasa_exito_job.index, y=tasa_exito_job.values, palette='viridis')

# 4. Personalización y etiquetas
plt.title('Tasa de Éxito de Suscripción por Profesión (%)', fontsize=16, fontweight='bold')
plt.xlabel('Profesión', fontsize=12)
plt.ylabel('Éxito en Suscripción (%)', fontsize=12)
plt.xticks(rotation=45) # Rotamos las etiquetas para que sean legibles

# 5. Añadir el valor exacto sobre cada barra para mayor precisión
for i, value in enumerate(tasa_exito_job.values):
    plt.text(i, value + 0.5, f'{value:.1f}%', ha='center', fontsize=10)

plt.tight_layout()
plt.show()
```
